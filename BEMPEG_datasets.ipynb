# scRNAseq metastatic melanoma

### Load melanoma datasets 1/2 and change the cell anotations

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scanpy as sc
```

```python
df = pd.read_csv("data/expression/melanoma/GSE72056.txt.gz", compression='gzip', sep='\t', header=None)
```

```python
# Arrange data in the same way as before
df = df.transpose()
df[0] = df[0]+"__"+df[1]+"__"+df[2]+"__"+df[3]
df.set_index(0, inplace=True)
df = df.transpose()
df.set_index(df.columns[0], inplace=True)
df.columns = [col.replace('-','.') for col in df.columns]
```

```python
# Add immune cell types
Cell_data = pd.read_csv("data/expression/melanoma/Layer2_Bcell_scImmuCC_label.csv")[["barcodes","cell_type"]]
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/Layer2_DC_scImmuCC_label.csv")[["barcodes","cell_type"]]], axis=0)
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/Layer2_ILC_scImmuCC_label.csv")[["barcodes","cell_type"]]], axis=0)
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/Layer2_Macrophage_scImmuCC_label.csv")[["barcodes","cell_type"]]], axis=0)
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/Layer2_Monocyte_scImmuCC_label.csv")[["barcodes","cell_type"]]], axis=0)
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/Layer2_NK_scImmuCC_label.csv")[["barcodes","cell_type"]]], axis=0)
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/Layer3_CD4_scImmuCC_label.csv")[["barcodes","cell_type"]]], axis=0)
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/Layer3_CD8_scImmuCC_label.csv")[["barcodes","cell_type"]]], axis=0)
```

```python
def convert_log2(x):
    return np.log2((2**x-1)*10+1)
```

```python
# Add the patient number to the cell type
Cell_data["Patient"] = [cell.split("__")[-3] for cell in Cell_data["barcodes"]]
Cell_data["cell_type"] = Cell_data["cell_type"] + "__" + Cell_data["Patient"]
# Change the cell annotations
df = df[Cell_data["barcodes"]]
df.columns = Cell_data["cell_type"]
# Put data as float and leave only genes
df = df.astype('float').apply(convert_log2)
df = df.loc[df.index[3:]]
df.index.name="Genes"
```

```python
# Convert gene name to uniprot ID
df_prot = pd.read_excel("data/expression/uniprot_data.xlsx")
df_prot = df_prot[["Entry Name", "Gene Names"]].dropna()
df_prot["Gene Names"] = [gene_name.split(" ")[0].replace("/","").replace(".","").replace(" ","").replace("-","").upper() for gene_name in df_prot["Gene Names"]]
df.index = [gene_name.split(" ")[0].replace("/","").replace(".","").replace(" ","").replace("-","").upper() for gene_name in df.index]
df = df.loc[df.index.isin(df_prot["Gene Names"])]
df_prot = df_prot.loc[df_prot["Gene Names"].isin(df.index)]
df_prot.index = df_prot['Gene Names']
uniprot_dict = df_prot["Entry Name"].to_dict()
df.index = [uniprot_dict[gene_name] for gene_name in df.index]
```

```python
# Load second dataset immune cells melanoma
df2 = pd.read_csv('data/expression/melanoma/dataset2/scRNAseq_melanoma2_annotated.tsv.gz', sep='\t', compression='gzip')
df2.set_index('0', inplace=True)
```

```python
# Add immune cell types
Cell_data = pd.read_csv("data/expression/melanoma/dataset2/Layer2_Bcell_scImmuCC_label.csv")[["barcodes","cell_type"]]
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/dataset2/Layer2_DC_scImmuCC_label.csv")[["barcodes","cell_type"]]], axis=0)
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/dataset2/Layer1_scImmuCC_label.csv")[["barcodes","cell_type"]].loc[pd.read_csv("data/expression/melanoma/dataset2/Layer1_scImmuCC_label.csv")[["barcodes","cell_type"]]["cell_type"]=="ILC"]], axis=0)
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/dataset2/Layer2_Macrophage_scImmuCC_label.csv")[["barcodes","cell_type"]]], axis=0)
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/dataset2/Layer2_Monocyte_scImmuCC_label.csv")[["barcodes","cell_type"]]], axis=0)
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/dataset2/Layer2_NK_scImmuCC_label.csv")[["barcodes","cell_type"]]], axis=0)
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/dataset2/Layer3_CD4_scImmuCC_label.csv")[["barcodes","cell_type"]]], axis=0)
Cell_data = pd.concat([Cell_data, pd.read_csv("data/expression/melanoma/dataset2/Layer3_CD8_scImmuCC_label.csv")[["barcodes","cell_type"]]], axis=0)
Cell_data["cell_type"] = Cell_data["cell_type"]+"__"+[cell.split("__Pre_")[1] for cell in Cell_data["barcodes"]]
Cell_data.index = [col.replace('.','-') for col in Cell_data["barcodes"]]
Cell_data = Cell_data["cell_type"].to_dict()
```

```python
# Convert gene name to uniprot ID
df_prot = pd.read_excel("data/expression/uniprot_data.xlsx")
df_prot = df_prot[["Entry Name", "Gene Names"]].dropna()
df_prot["Gene Names"] = [gene_name.split(" ")[0].replace("/","").replace(".","").replace(" ","").replace("-","").upper() for gene_name in df_prot["Gene Names"]]
df2.index = [gene_name.split(" ")[0].replace("/","").replace(".","").replace(" ","").replace("-","").upper() for gene_name in df2.index]
df2 = df2.loc[df2.index.isin(df_prot["Gene Names"])]
df_prot = df_prot.loc[df_prot["Gene Names"].isin(df2.index)]
df_prot.index = df_prot['Gene Names']
uniprot_dict = df_prot["Entry Name"].to_dict()
df2.index = [uniprot_dict[gene_name] for gene_name in df2.index]
```

```python
# Combine both datasets, change last number indicates the corresponding patient Xnumber of that cell type in ge column names so t export it
df = df[~df.index.duplicated(keep='first')]
df2 = df2[~df2.index.duplicated(keep='first')]
df_final = pd.concat([df, df2], axis=1)
df_final = df_final.fillna(0)
df_final.columns = [col.split(".")[0] for col in df_final.columns]
# Iterate through the columns and rename repeated columns
col_counts = {}
new_columns = []
for col in df_final.columns:
    if col in col_counts:
        col_counts[col] += 1
        new_columns.append(f"{col}__{col_counts[col]}")
    else:
        col_counts[col] = 1
        new_columns.append(f"{col}__{col_counts[col]}")

# Rename the columns of the DataFrame
df_final.columns = new_columns
df_final.to_csv('data/expression/melanoma/melanoma_comb.tsv.gz', sep='\t', compression='gzip')
```

## scGen for batch effect correction

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import use
use('Agg')
import warnings
warnings.filterwarnings('ignore')
plt.rc('xtick', labelsize=14)
plt.rc('ytick', labelsize=14)
import scanpy as sc
from anndata import AnnData
import scgen
import decoupler as dc
import matplotlib as mpl
import glasbey
```

```python
# Load scRNAseq melanoma data
df_final = pd.read_csv('data/expression/melanoma/melanoma_comb.tsv.gz',sep='\t', compression='gzip', index_col=0)
adata = AnnData(df_final.transpose())
batch_list = list(adata.obs.index.str.contains("__P"))
batch_list2 = []
for i in range(0,len(batch_list)):
    if batch_list[i]:
        batch_list[i]='2'
    else:
        batch_list[i]='1'
adata.obs["batch"] = batch_list
adata.obs["cell_type"] = [cell.split("__")[0] for cell in adata.obs.index]
```

### Do batch correction of melanoma dataset

```python
# Load trained model
model = scgen.SCGEN.load("data/expression/melanoma/models/model_batch_removal.pt",adata)
# Batch_corrected data
corrected_adata = model.batch_removal()
# Compute distances in the PCA space, and find cell neighbors
sc.pp.neighbors(corrected_adata, n_neighbors=10, n_pcs=40)
# Generate UMAP features
sc.tl.umap(corrected_adata)
# Run leiden clustering algorithm
sc.tl.leiden(corrected_adata)
# Save the batch corrected data on a dataframe
df_final = corrected_adata.to_df().transpose()
```

```python
# Generate umap of batch corrected data
fig,ax=plt.subplots(1,1, dpi=400)
sc.pl.umap(corrected_adata, color=['batch'], show=False,ax=ax)
ax.set_title("")
ax.spines["bottom"].set_linewidth(4)
ax.spines["left"].set_linewidth(4)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.xaxis.set_tick_params(width=5, length=10)
ax.yaxis.set_tick_params(width=5, length=10)
ax.tick_params(axis='x', labelsize=25)
ax.tick_params(axis='y', labelsize=25)
plt.ylabel('UMAP2', fontsize=15)
plt.xlabel('UMAP1', fontsize=15)
plt.savefig('figures/melanoma/UMAP_corrected_batch.pdf', bbox_inches='tight', transparent=True)
plt.close()

fig,ax=plt.subplots(1,1, dpi=400)
sc.pl.umap(corrected_adata, color=['cell_type'], show=False,ax=ax, palette=palette)
ax.set_title("")
ax.spines["bottom"].set_linewidth(4)
ax.spines["left"].set_linewidth(4)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.xaxis.set_tick_params(width=5, length=10)
ax.yaxis.set_tick_params(width=5, length=10)
ax.tick_params(axis='x', labelsize=25)
ax.tick_params(axis='y', labelsize=25)
plt.ylabel('UMAP2', fontsize=15)
plt.xlabel('UMAP1', fontsize=15)
plt.savefig('figures/melanoma/UMAP_corrected_cell_type.pdf', bbox_inches='tight', transparent=True)
plt.close()

fig,ax=plt.subplots(1,1, dpi=400)
sc.pl.umap(corrected_adata, color=['leiden'], show=False,ax=ax)
ax.set_title("")
ax.spines["bottom"].set_linewidth(4)
ax.spines["left"].set_linewidth(4)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.xaxis.set_tick_params(width=5, length=10)
ax.yaxis.set_tick_params(width=5, length=10)
ax.tick_params(axis='x', labelsize=25)
ax.tick_params(axis='y', labelsize=25)
plt.ylabel('UMAP2', fontsize=15)
plt.xlabel('UMAP1', fontsize=15)
plt.savefig('figures/melanoma/UMAP_corrected_leiden.pdf', bbox_inches='tight', transparent=True)
plt.close()
```

# Load BEMPEG simulation data and generate UMAP

```python
obs_adata = corrected_adata.obs.copy() # Copy observables of Anndata object
```

```python
# Load data and separate it on 3 groups, non-IL2 target cells and IL2 target activated and unactivated cells
df = pd.read_csv("results/BEMPEG_trials/simulations_BEMPEG_tumor_IL2s.csv")
df_high = df.loc[df["IL0_1"]==1e-4][["Cell_type","RB0","STAT50","Result"]]
df_treatment = df.loc[df["IL0_1"]!=1e-4][["Cell_type","RB0","STAT50","Result"]]
# Separate cells into 3 groups
df_treatment["Abs. activ."] = np.log10(df_treatment["Result"].values)
df_treatment["Activ. pot."] = df_treatment["Result"].values/df_high["Result"].values*100
# Get simulation data into anndata object to create UMAPs
corrected_adata.obs = pd.merge(obs_adata, df_treatment[["Cell_type","Abs. activ.","Activ. pot."]], how='left', left_index=True, right_on='Cell_type')
corrected_adata.obs.set_index('Cell_type', inplace=True)
```

### Subclustering of Tregs into PD1+ or not

```python
# Separate Tregs into PD1+ or not
df_cells = df_final[df_final.columns[df_final.columns.str.contains("Treg")]]
adata_cells = AnnData(df_cells.transpose())

# Compute distances in the PCA space, and find cell neighbors
sc.pp.neighbors(adata_cells, n_neighbors=10, n_pcs=40)
# Generate UMAP features
sc.tl.umap(adata_cells)
# Run leiden clustering algorithm
sc.tl.leiden(adata_cells)

# Do violin plot of PD1 expression of Tregs in the leiden clusters
medians = []
conf_int = []
for i in adata_cells.obs["leiden"].drop_duplicates().values:
    gene_expression = df_cells.loc[["PDCD1_HUMAN"]][df_cells.loc[["PDCD1_HUMAN"]].columns[adata_cells.obs["leiden"]==i]].values
    conf_int.append(1.96 * (np.std(gene_expression) / np.sqrt(len(gene_expression))))
    medians.append(np.median(gene_expression))
conf_int = np.array(conf_int)
medians = np.array(medians)
fig,ax=plt.subplots(1,1, dpi=400)
sc.pl.violin(adata_cells, keys=["PDCD1_HUMAN"], groupby='leiden', show=False,ax=ax)
plt.scatter(adata_cells.obs["leiden"].drop_duplicates().values.astype(np.int32),medians, color='black')
plt.errorbar(adata_cells.obs["leiden"].drop_duplicates().values.astype(np.int32), medians, conf_int, linewidth=1.2, fmt='o', markersize=4, color='black' , ecolor='black', elinewidth=1, capsize=5, barsabove=True)
ax.set_title("")
ax.spines["bottom"].set_linewidth(4)
ax.spines["left"].set_linewidth(4)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.xaxis.set_tick_params(width=5, length=10)
ax.yaxis.set_tick_params(width=5, length=10)
ax.tick_params(axis='x', labelsize=15)
ax.tick_params(axis='y', labelsize=15)
plt.ylabel("PDCD1_HUMAN", fontsize=15)
plt.xlabel('Leiden', fontsize=15)
plt.savefig('figures/melanoma/violin_Tregs.pdf', bbox_inches='tight', transparent=True)
plt.close()

# Put PD1 data as obsm so that we can later do a multivariate 
adata_cells.obsm['ora_estimate'] = df_cells.transpose()[["PDCD1_HUMAN"]]
# To use scanpy functions, generate an anndata object with the enrichment results as main data
acts = dc.get_acts(adata_cells, obsm_key='ora_estimate')
# Remove inf and set them to the maximum value observed for pvals=0
acts_v = acts.X.ravel()
max_e = np.nanmax(acts_v[np.isfinite(acts_v)])
acts.X[~np.isfinite(acts.X)] = max_e

# Welch T-test 
df = dc.rank_sources_groups(acts, groupby='leiden', reference='rest', method='t-test_overestim_var')

# Get clusters with increased PD1 expression and p_value < 0.01
cells_pos = pd.DataFrame(acts.obs['leiden'])[acts.obs['leiden'].isin(df.loc[df["pvals_adj"]<0.01].loc[df.loc[df["pvals_adj"]<0.01,"meanchange"]>0,"group"].values)].index
col_new = []
for col in df_final.columns:
    if col in cells_pos:
        col_new.append(col.split('__')[0]+'_PD1+__'+col.split('__')[1]+'__'+col.split('__')[2])
    else:
        col_new.append(col)
df_final.columns = col_new
```

### Subclustering of CD8 Tcells into cytotoxic or not

```python
# Cells we want to subcluster
cells_subcluster = ['CD8_Effector_memory','CD8_Central_memory','CD8_Exhausted']
# Generate a Cytotoxicity marker 
markers = pd.DataFrame()
markers["genesymbol"] = ["GRAA_HUMAN","GRAB_HUMAN","GRAH_HUMAN","GRAK_HUMAN","GRAM_HUMAN","PERF_HUMAN","TCF7_HUMAN","TOX_HUMAN","B3GA1_HUMAN"]
markers["cell_type"] = ["effector+","effector+","effector+","effector+","effector+","effector+","effector+","effector+","effector+"]

for cell in cells_subcluster:
    # Generate anndata file with only th data of the cell type
    df_cells = df_final[df_final.columns[df_final.columns.str.contains(cell)]]
    adata_cells = AnnData(df_cells.transpose())

    # Compute distances in the PCA space, and find cell neighbors
    sc.pp.neighbors(adata_cells, n_neighbors=10, n_pcs=40)
    # Generate UMAP features
    sc.tl.umap(adata_cells)
    # Run leiden clustering algorithm
    sc.tl.leiden(adata_cells)

    # Enrichment analyis on the cytotoxic subtype generated by the markers
    dc.run_ora(mat=adata_cells,net=markers,source='cell_type',target='genesymbol',min_n=3,verbose=True,use_raw=False)

    # Do violin plot of the enrichment with Over Representation Analysis (ORA) in CD8 cell types in the leiden clusters
    adata_cells.obs['ora_estimate'] = adata_cells.obsm['ora_estimate']
    medians = []
    conf_int = []
    for i in adata_cells.obs["leiden"].drop_duplicates().values:
        gene_expression = adata_cells.obsm['ora_estimate'].loc[adata_cells.obs["leiden"].index[adata_cells.obs["leiden"]==i],"effector+"].values
        conf_int.append(1.96 * (np.std(gene_expression) / np.sqrt(len(gene_expression))))
        medians.append(np.median(gene_expression))
    conf_int = np.array(conf_int)
    medians = np.array(medians)
    fig,ax=plt.subplots(1,1, dpi=400)
    sc.pl.violin(adata_cells, keys=['ora_estimate'], groupby='leiden', show=False,ax=ax)
    plt.scatter(adata_cells.obs["leiden"].drop_duplicates().values.astype(np.int32),medians, color='black')
    plt.errorbar(adata_cells.obs["leiden"].drop_duplicates().values.astype(np.int32), medians, conf_int, linewidth=1.2, fmt='o', markersize=4, color='black' , ecolor='black', elinewidth=1, capsize=5, barsabove=True)
    ax.set_title("")
    ax.spines["bottom"].set_linewidth(4)
    ax.spines["left"].set_linewidth(4)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.xaxis.set_tick_params(width=5, length=10)
    ax.yaxis.set_tick_params(width=5, length=10)
    ax.tick_params(axis='x', labelsize=15)
    ax.tick_params(axis='y', labelsize=15)
    plt.ylabel("ORA estimate", fontsize=15)
    plt.xlabel('Leiden', fontsize=15)
    plt.savefig('figures/melanoma/violin_'+cell+'.pdf', bbox_inches='tight', transparent=True)
    plt.close()

    # To use scanpy functions, generate an anndata object with the enrichment results as main data
    acts = dc.get_acts(adata_cells, obsm_key='ora_estimate')
    # Remove inf and set them to the maximum value observed for pvals=0
    acts_v = acts.X.ravel()
    max_e = np.nanmax(acts_v[np.isfinite(acts_v)])
    acts.X[~np.isfinite(acts.X)] = max_e

    # Welch T-test 
    df = dc.rank_sources_groups(acts, groupby='leiden', reference='rest', method='t-test_overestim_var')

    # Get clusters with increased ORA estimate and p_value < 0.01
    cells_pos = pd.DataFrame(acts.obs['leiden'])[acts.obs['leiden'].isin(df.loc[df["pvals_adj"]<0.01].loc[df.loc[df["pvals_adj"]<0.01,"meanchange"]>0,"group"].values)].index
    col_new = []
    for col in df_final.columns:
        if col in cells_pos:
            col_new.append(col.split('__')[0]+'_cytotoxic__'+col.split('__')[1]+'__'+col.split('__')[2])
        else:
            col_new.append(col)
    df_final.columns = col_new
    print(cell+': '+str(acts.obs['ora_estimate'].mean()))
```
